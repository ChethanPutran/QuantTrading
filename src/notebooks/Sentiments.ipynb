# 📈 Smart Financial News & Trading Assistant

An intelligent real-time financial assistant that:
- Fetches live news.
- Extracts relevant stocks.
- Predicts intraday movements.
- Suggests trades using technical indicators.
- Alerts you to breaking news that might impact your trades.

---

## 🔧 Features

### 1. 📰 Live Financial News Analysis
- Pulls the latest financial news from the internet (Google News, Twitter, or NewsAPI.org).
- Extracts **company names or stock symbols** using LLM-based Named Entity Recognition.
- Predicts whether a stock is likely to **move significantly intraday** based on news context.

---

### 2. 👤 User Interaction
- Displays a list of **stocks mentioned** in news headlines.
- Lets user **select stocks** they’re interested in trading today.

---

### 3. 📊 Intraday Charts & Trading Suggestions
- Shows **live intraday charts** (candlestick/line).
- Suggests **Buy / Sell / Hold** decisions based on:
  - **Candle patterns** (e.g., Doji, Engulfing, Hammer)
  - **Technical indicators** (e.g., RSI, MACD, Moving Averages)

---

### 4. 🔔 Real-Time Breaking News Alerts
- Monitors selected stocks continuously.
- Uses LLM to detect **positive/negative** breaking news sentiment.
- Sends **immediate alerts** when something impactful hits.

---

## 🧠 Technologies Used

### 📌 Core Technologies
- **Python**: Backend + LLM integration
- **Streamlit / Gradio (MVP)** or **React.js + FastAPI (Scalable)**: Web interface
- **Socket.IO / Websockets**: Real-time news & price updates
- **Plotly / mplfinance**: For charting

### 🗂️ Data Sources
- **News**: Google News API, Twitter scraping, NewsAPI.org
- **Stock Data**: Yahoo Finance, Alpha Vantage, or Polygon.io

### 🤖 LLM Capabilities
- **Summarizing** news headlines
- **Sentiment analysis** of financial news
- **Impact prediction** (positive/negative for stocks)
- Powered by OpenAI, Gemini, Claude, or LLaMA-based models
